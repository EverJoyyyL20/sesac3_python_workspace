```python
class Bicycle():
    def __init__(self,wheel_size,color):
        self.wheel_size=wheel_size
        self.color=color
    def move(self,speed):
        print("자전거: 시속 {0}킬로미터로 전지".format(speed))
    def turn(self,direction):
        print("자전거: {0}회전".format(direction))
    def stop(self):
        print("자전거({0},{1}):정지 ".format(self.wheel_size,self.color))
```

```python
my_bicycle=Bicycle(26,'black')

my_bicycle.move(30)
my_bicycle.turn('좌')
my_bicycle.stop()
```

```python
class Car():
    instance_count=0

    def __init__(self,size,color):
        self.size=size
        self.color=color
        Car.instance_count=Car.instance_count+1
        print("자동치 객체의 수:{0}".format(Car.instance_count))
    def move(self):
        print("자동차({0} & {1})가 움직입니다.".format(self.size,self.color))
   
```

```python
class Car():
    instance_count=0

    def __init__(self,size,color):
        self.size=size
        self.color=color
        Car.instance_count=Car.instance_count+1
        print("자동차 객체의 수:{0}".format(Car.instance_count))

    def move(self,speed):
        self.speed=speed
        print("자동차({0} & {1})가".format(self.size,self.color),end='')
        print("시속 {0} 킬로미터로 전진".format(self.speed))

    def auto_cruise(self):
        print("자율주행모드")
        self.move(self.speed)
```

```python
car1=Car("small","red")
car2=Car("big","green")

car1.move(80)
car2.move(100)

car1.auto_cruise()
car2.auto_cruise()
```

```python
class Car():


    @staticmethod
    def check_type(model_code):
        if(model_code>20):
            print("이 자동차는 전기차 입니다.")
        elif(10<= model_code<20):
            print("이 자동차는 가솔린차입니다.")
        else:
            print("이 자동차는디젤차입니다.") 
```

```python
Car.check_type(25)
Car.check_type(2)
```

```python
class Car():
    instance_count=0

    def __init__(self,size,color):
        self.size=size
        self.color=color
        Car.instance_count=Car.instance_count+1

    @classmethod
    def count_instance(cls):
        print("자동차 객체의 개수:{0}".format(cls.instance_count))
        
```

```python
Car.count_instance()

car1=Car("small","red")
Car.count_instance()


car2=Car("big","green")
Car.count_instance()
```
